```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

# Ensure project root is in path
sys.path.append(os.path.abspath('..'))
sns.set_theme(style='whitegrid')
```

## Step 3: Dataset Collection & Understanding

```python
df_raw = pd.read_csv('../data/poll_data.csv')
print('🔍 Preview of Raw Poll Data:')
display(df_raw.head())

print('\n📊 Dataset Info:')
print(df_raw.info())

print('\n❓ Missing Values per Column:')
print(df_raw.isnull().sum())

print('\n🧾 Unique Poll Options (Before Cleaning):')
print(df_raw['Preferred Tool'].unique())
```

## Step 4: Data Cleaning & Preprocessing

```python
from src.data_cleaner import clean_poll_data

df, cleaning_stats = clean_poll_data(
    input_path='../data/poll_data.csv',
    output_path='../data/cleaned_poll_data.csv'
)
print('✅ Cleaning Audit Summary:', cleaning_stats)
display(df.head())
```

## Step 5: Statistical Aggregation & Exploratory Data Analysis (EDA)

```python
from src.analyzer import compute_vote_share_summary, compute_regional_breakdown, compute_demographic_breakdown

vote_summary = compute_vote_share_summary(df)
print('📈 Option-Wise Vote Share Summary:')
display(vote_summary)

regional_summary = compute_regional_breakdown(df)
print('🌍 Regional Breakdown:')
display(regional_summary)
```

```python
# 1. Bar Chart - Most Preferred Tools
plt.figure(figsize=(8, 5))
order = df['Preferred Tool'].value_counts().index
sns.countplot(x='Preferred Tool', hue='Preferred Tool', data=df, order=order, palette='Set2', legend=False)
plt.title('Most Preferred Tools (Vote Count)')
plt.xlabel('Tool')
plt.ylabel('Number of Votes')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# 2. Histogram - Satisfaction Ratings Distribution
plt.figure(figsize=(6.5, 4.2))
sns.histplot(df['Satisfaction (1-5)'], bins=5, kde=True, color='skyblue')
plt.title('Satisfaction Rating Distribution (1-5)')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# 3. Line Plot - Daily Poll Submissions
daily = df.groupby('Date').size()
plt.figure(figsize=(9, 4))
daily.plot(kind='line', marker='o', color='#2563EB')
plt.title('Daily Poll Submissions Over 30-Day Window')
plt.xlabel('Date')
plt.ylabel('Number of Responses')
plt.xticks(rotation=30)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 4. Word Cloud - Common Words in Feedback
text = ' '.join(df['Feedback'].astype(str).tolist())
wordcloud = WordCloud(width=800, height=400, background_color='white', colormap='tab10').generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Common Words in Open-Ended Feedback')
plt.show()
```

```python
# 5. Boxplot - Feedback Length vs. Satisfaction
plt.figure(figsize=(7, 5))
sns.boxplot(x='Satisfaction (1-5)', y='Feedback Length', hue='Satisfaction (1-5)', data=df, palette='coolwarm', legend=False)
plt.title('Feedback Length by Satisfaction Level')
plt.tight_layout()
plt.show()
```
